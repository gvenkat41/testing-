```python
nums =[1,2,3,4,5,6,7,8,9,4,5,6,2,4,6]
```

```python
duplicates=[]
def remove_duplicates():
    for num in nums:
        if nums.count(num)>1 and num  not in duplicates:
            duplicates.append(num)
    return duplicates
print(remove_duplicates())
```
